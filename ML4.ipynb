```python
import math
import statistics

print("=== Python 'math' Library Examples ===")
print(f"Value of pi: {math.pi}")
print(f"Value of e: {math.e}")
print(f"Sine of 90 degrees (pi/2 radians): {math.sin(math.pi / 2)}")
print(f"Square root of 144: {math.sqrt(144)}")
print(f"2 raised to the power of 10: {math.pow(2, 10)}")
print(f"Natural log (base e) of 10: {math.log(10):.4f}")
print(f"Log base 10 of 100: {math.log10(100)}")
number = 4.3
print(f"Ceiling of {number}: {math.ceil(number)}")
print(f"Floor of {number}: {math.floor(number)}")
print(f"Factorial of 5: {math.factorial(5)}")
print(f"Greatest Common Divisor (GCD) of 24 and 36: {math.gcd(24, 36)}")

print("\n=== Python 'statistics' Library Examples ===")
data_list = [2, 4, 4, 4, 5, 5, 7, 9, 20]
print(f"Dataset: {data_list}")
print(f"Arithmetic Mean: {statistics.mean(data_list):.4f}")
print(f"Median (middle value): {statistics.median(data_list)}")
print(f"Mode (most common value): {statistics.mode(data_list)}")
print(f"Geometric Mean: {statistics.geometric_mean(data_list):.4f}")
print(f"Variance: {statistics.variance(data_list):.4f}")
print(f"Standard Deviation: {statistics.stdev(data_list):.4f}")
```
