```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.manifold import Isomap
from sklearn.preprocessing import StandardScaler

# Generate sample data
np.random.seed(42)
data = {
    'Category': np.random.choice(['A', 'B', 'C', 'D'], size=100),
    'Value1': np.random.randint(10, 100, size=100),
    'Value2': np.random.uniform(5.0, 50.0, size=100),
    'Feature1': np.random.randn(100) * 10 + 50,
    'Feature2': np.random.randn(100) * 5 + 20,
    'Feature3': np.random.randn(100) * 2 + 10
}
df = pd.DataFrame(data)

display(df.head())
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Category', y='Value1', data=df, palette='Set2', hue='Category', legend=False)
plt.title('Distribution of Value1 by Category (Box Plot)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
corr = df.select_dtypes(include=[np.number]).corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Feature Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.violinplot(x='Category', y='Value2', data=df, palette='Pastel1', hue='Category', legend=False)
plt.title('Density and Distribution of Value2 by Category')
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Mapping categories to colors
colors = {'A': 'red', 'B': 'blue', 'C': 'green', 'D': 'orange'}

ax.scatter(df['Feature1'], df['Feature2'], df['Feature3'],
           c=df['Category'].map(colors), s=50, alpha=0.6)

ax.set_xlabel('Feature 1')
ax.set_ylabel('Feature 2')
ax.set_zlabel('Feature 3')
plt.title('3D Scatter Plot of Numerical Features')
plt.show()
```
